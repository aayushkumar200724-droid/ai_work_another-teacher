```python
# IMAGE EXTRACT

import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python
img =cv2.imread('messi.jpg.jpeg',cv2.IMREAD_COLOR)
```

```python
print(type(img))
print(img.shape)
print(img.dtype)
```

```python
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))  # file read krna
```

```python
for i in range(30):     # 0 SE 30 TAK LOOP CHALEGA
    for j in range(30):    # SAME YAHA BHI LOOP CHALEGA
        img[50+i,235+j]=(0,255,0)      # YE SARE PIXEL PR JAKR DOT SE BANA HAI ONLY DOT SE , DOT SE BANA HAI ISLIYE LOOP CHALAYA HAI
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
```

# EXTRACT THE BALL


```python
ball=img[280:340,330:390]
plt.imshow(cv2.cvtColor(ball,cv2.COLOR_BGR2RGB))
```

# EK RANDOM PIXEL GENERATE KRKE VAHA PR ALAG ALAG BALL BANA NI HAI


```python
top_left_x=np.random.randint(0,480,5)  # (start, last , how many parts)
top_left_y=np.random.randint(0,280,5)
for x,y in zip(top_left_x,top_left_y):
    img[y:y+60 , x:x+60]=ball
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
```
