```python
# IMAGE TRANSLATION AND ROTATION
import cv2
import numpy as np
from matplotlib import pyplot as plt
```

```python
image = cv2.imread('input.jpg.jpeg')
```

```python
height,width =image.shape[:2]
quarter_height, quarter_width = height/4, width/4
```

```python
print("height.shape ", height)
print("quarter_height.shape ", quarter_height)
```

```python
print("height.shape ", height)
print("quarter_height.shape ", quarter_width)
```

```python
# T is our translation matrix
T=np.float32([[1,0,quarter_width],
             [0,1,quarter_height]])
# let's take a look at T
print(T)
```

```python
# WE use warpAffine to transform the image using the matrix,
img_translation = cv2.warpAffine(image,T,(width,height))
print(img_translation.shape)
```

```python
plt.imshow(cv2.cvtColor(img_translation,cv2.COLOR_BGR2RGB))
plt.show('Translation')

plt.show()
```
