```python
import qrcode
```

```python
# data to encode
data=["Nittin","01-01-2006","O+",7509656437]
# creating an instance of qrcode class
qr=qrcode.QRCode()

# adding data to the instance "qr"
qr.add_data(data)

qr.make(fit=True)
image=qr.make_image()
image.save("nittin_qr.png")
```

```python
import cv2
d=cv2.QRCodeDetector()
text, _, _ =d.detectAndDecode(cv2.imread("nittin_qr.png"))
print("Decode Text is ",text)
```
