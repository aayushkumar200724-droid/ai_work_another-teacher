```python
# Fram creation

import cv2
import  os
```

```python
cam=cv2.VideoCapture("C:/Users/nitti/Downloads/❤️#trendingnow #fyp #explore #real #love [ BeYourself RealMe Authenticity KnowYourWorth SelfLove.mp4")
```

```python
try:
    #create a folder named data
    if not os.path.exists("./accident_data/frames"):
        os.makedirs("./accident_data/frames")
# if not created then raise error
except OSError:
    print("Error:Creating Directory of frames ")
```

```python
#frame
currentframe=0

while(True):
    # reading fram frame
    ret,frame=cam.read()

    if ret:
        # if video is still left continue creating images
        name="./accident_data/frames/frame" + str(currentframe) + ".jpg"
        print("Creating..."+name)

        # writing the extracted images
        cv2.imwrite(name,frame)

        # increasinf counter so that it will
        # show how many frames are created
        currentframe+=1
    else:
        break

# release all space and windows once done
cam.release()
cv2.destroyAllWindows()
```
